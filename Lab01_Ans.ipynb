### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
import math
a, b, c = map(int, input().split())
def area(a, b, c):
    if a+b<=c or a+c<=b or b+c<=a or a<=0 or b<=0 or c<=0:
        raise ValueError("The sides cannot form a triangle")
    else:
        p=(a+b+c)/2
        return math.sqrt(p*(p-a)*(p-b)*(p-c))
print(area(a, b, c))
```

### Problem 7

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
x = ord(input())
n = int(input())
def p(x, n):
    if 65 <= x <= 90:
        position = x - ord("A")
        new_position = (position + n) % 26
        return chr(new_position + ord("A"))
    elif 97 <= x <= 122:
        position = x - ord("a")
        new_position = (position + n) % 26
        return chr(new_position + ord("a")) 
    else:
        return None
print(p(x, n))

    
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
u = input()
def convert(u):
    if len(u) == 1 and u.isupper():
        return u.lower()
    else:
        raise ValueError("The input is not a single character or an uppercase")
print(convert(u))
```
